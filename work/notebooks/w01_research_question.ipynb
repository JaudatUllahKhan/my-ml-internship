## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os, sys, subprocess
import pandas as pd
import numpy as np

# Ensure working directory is set correctly in Colab
IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)
elif os.path.basename(os.getcwd()) == "notebooks":
    os.chdir("../..")

# Load starter dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Compute supporting real numbers for Lane 1
total_pages = len(df)
declining_pages = (df['trend_direction'].str.lower() == 'down').sum()
declining_pct = (declining_pages / total_pages) * 100

high_exposure_declining = df[(df['trend_direction'].str.lower() == 'down') & (df['impressions_90d'] > 1000)]
high_exposure_pct = (len(high_exposure_declining) / total_pages) * 100

avg_days_stale = df[df['trend_direction'].str.lower() == 'down']['days_since_last_update'].mean()

# Print findings
print(f"Total Pages Analyzed: {total_pages:,}")
print(f"1. Overall Declining Pages: {declining_pages:,} ({declining_pct:.1f}% of total pages)")
print(f"2. High-Exposure Declining Pages (>1,000 impressions): {len(high_exposure_declining):,} ({high_exposure_pct:.1f}% of total dataset)")
print(f"3. Mean Stale Time for Declining Content: {avg_days_stale:.1f} days since last update")
```
